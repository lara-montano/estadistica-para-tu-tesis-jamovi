```python
import pandas as pd, pingouin as pg, seaborn as sns
df = pd.read_csv("https://raw.githubusercontent.com/lara-montano/estadistica-para-tu-tesis-jamovi/main/datos/datos.csv")
print(df.shape)
df.head()
```

## Descriptiva y normalidad

```python
df["desempeno_post"].describe()
```

```python
df.groupby("grupo")["desempeno_post"].agg(["mean","std","count"])
```

```python
pg.normality(df, dv="desempeno_post", group="grupo")
```

## Prueba t (independiente y pareada)

```python
a = df.loc[df.grupo=="Método A","desempeno_post"]
c = df.loc[df.grupo=="Control","desempeno_post"]
pg.ttest(a, c)   # T, dof, p, Cohen-d, IC95% de la diferencia
```

```python
pg.ttest(df["desempeno_post"], df["desempeno_pre"], paired=True)
```

## ANOVA + post-hoc

```python
pg.anova(data=df, dv="desempeno_post", between="grupo")
```

```python
pg.pairwise_tukey(data=df, dv="desempeno_post", between="grupo")
```

## Correlación y regresión

```python
pg.corr(df["horas_practica"], df["desempeno_post"])  # Pearson
```

```python
pg.corr(df["satisfaccion"], df["desempeno_post"], method="spearman")
```

```python
pg.linear_regression(df[["horas_practica","edad","motivacion"]].dropna(),
                     df.loc[df[["horas_practica","edad","motivacion"]].dropna().index,"desempeno_post"])
```

## Chi-cuadrado y fiabilidad (alfa de Cronbach)

```python
exp, obs, stats = pg.chi2_independence(df, x="grupo", y="aprobado")
stats
```

```python
pg.cronbach_alpha(data=df[["util_1","util_2","util_3","util_4","util_5"]].dropna())
```
